```python
import pandas as pd    # importing Lib.for analysis
import numpy as np   # importing Lib. for numeric computations
```

```python
data_set = pd.read_csv('spam.csv',encoding='latin-1')   # Fetching the dataset
df= pd.DataFrame(data_set)       #converting dataset into dataframe
df
```

```python
import re       #importing Regular Expression module for text cleaning
def clean_num(text):
    text = re.sub(r'\d+', ' NUM ', text)            #\d matches decimal
    return text
df['input']= df['v2'].apply(clean_num)
df.drop(['v2'], axis=1,inplace= True)
```

```python
# Vectorization (INPUT DATASET ) using Term frequency Inverse document frequency
from sklearn.feature_extraction.text import TfidfVectorizer
tf_idf = TfidfVectorizer(stop_words='english',max_features= 3000)
X = tf_idf.fit_transform(df['input']).toarray()
X                           #INput Feature
```

```python
# output feature
y= df.iloc[:,0]
#converting string into number for ml algo
df['v1']= df['v1'].map({'spam':1 , 'ham': 0})

```

```python
#Splitting the dataset into train and test data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.21, random_state=42)

print(len(X_train), len(y_train))   #checking whether the length is same or not
```

```python
from sklearn.naive_bayes import MultinomialNB
nai_ba = MultinomialNB()
nai_ba.fit(X_train,y_train)
```

```python
y_pred = nai_ba.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
f"accuracy using naive bayes is :{accuracy_score(y_test, y_pred)}"
```

```python
#save the model
import pickle
filename = 'save_model.sav'
tf_vec= 'vectorizer.pkl'
pickle.dump(nai_ba,open(filename,"wb"))

# save the TFIDF model also
pickle.dump(tf_idf, open(tf_vec, 'wb'))

#load the model
load_model = pickle.load(open(filename,"rb"))
```
